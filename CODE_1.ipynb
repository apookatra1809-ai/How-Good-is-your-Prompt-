```python
# ORIGINAL
ORIGINAL_INPUT = "/kaggle/input/datasets/sharathraoram/123dataaaaa/phi_original_checkpoint_1024.csv"

# VAGUE
VAGUE_INPUT = "/kaggle/input/datasets/sharathraoram/123dataaaaa/phi_vague_checkpoint_1024.csv"

ORIGINAL_OUTPUT = "phi_original_lexical_diversity.csv"

VAGUE_OUTPUT = "phi_vague_lexical_diversity.csv"
```

```python
def compute_mtld(text):

    if pd.isna(text):
        return np.nan

    text = str(text).strip()

    if len(text.split()) < 20:
        return np.nan

    try:
        lex = LexicalRichness(text)
        return lex.mtld()
    except:
        return np.nan
```

```python
def process_file(input_file, output_file):

    df = pd.read_csv(input_file)

    print(f"Processing {len(df)} responses...")

    df["mtld"] = df["response"].progress_apply(compute_mtld)

    df.to_csv(output_file, index=False)

    print(f"Saved to {output_file}")

    return df
```

```python
original_df = process_file(

    ORIGINAL_INPUT,

    ORIGINAL_OUTPUT

)
```

```python
vague_df = process_file(
    VAGUE_INPUT,
    VAGUE_OUTPUT
)
```
