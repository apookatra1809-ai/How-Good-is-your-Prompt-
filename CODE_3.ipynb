```python
import pandas as pd
import numpy as np

# ============================
# FILE PATHS
# ============================

files = [
    "/kaggle/input/datasets/sharathraoram/task-alignment/mistral_original_task_alignment.csv",
    "/kaggle/input/datasets/sharathraoram/task-alignment/mistral_vague_task_alignment.csv",
    "/kaggle/input/datasets/sharathraoram/task-alignment/phi_original_task_alignment.csv",
    "/kaggle/input/datasets/sharathraoram/task-alignment/phi_vague_task_alignment.csv",
    "/kaggle/input/datasets/sharathraoram/task-alignment/qwen_original_task_alignment.csv",
    "/kaggle/input/datasets/sharathraoram/task-alignment/qwen_vague_task_alignment.csv"
]

# ============================
# TASK ALIGNMENT SCORING
# ============================

def task_alignment_score(f1):

    if pd.isna(f1):
        return np.nan

    if f1 <= 0.8234:
        return 1

    elif f1 <= 0.8411:
        return 2

    elif f1 <= 0.8598:
        return 3

    elif f1 <= 0.8934:
        return 4

    else:
        return 5

# ============================
# READ + SCORE
# ============================

all_data = []

for file in files:

    df = pd.read_csv(file)

    df["task_alignment_score"] = df["bertscore_f1"].apply(task_alignment_score)

    all_data.append(df)

# ============================
# COMBINE
# ============================

final_df = pd.concat(all_data, ignore_index=True)

# ============================
# KEEP REQUIRED COLUMNS
# ============================

final_df = final_df[
    [
        "prompt_id",
        "model",
        "prompt_type",
        "bertscore_f1",
        "task_alignment_score"
    ]
]

# ============================
# SAVE
# ============================

OUTPUT_FILE = "task_alignment_scores.csv"

final_df.to_csv(OUTPUT_FILE, index=False)

print("Saved Successfully:", OUTPUT_FILE)

print(final_df.shape)

display(final_df.head(20))
```
